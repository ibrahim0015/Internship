```python
import pandas as pd
from sklearn.preprocessing import StandardScaler
from sklearn.compose import ColumnTransformer
from sklearn.preprocessing import OneHotEncoder
from sklearn.cluster import KMeans
from sklearn.pipeline import Pipeline
import matplotlib.pyplot as plt
```

```python
df = pd.read_csv('Mall_Customers.csv')
X = df.drop(['Gender','Age','CustomerID'],axis=1)
scaler = StandardScaler()
```

```python
# elbow method to figure out number of clusters
wcss = []
for i in range(1,11):
    pipeline = Pipeline([
        ('scaler',StandardScaler()),
        ('kmeans',KMeans(n_clusters=i,random_state=42))
    ])
    
    pipeline.fit_predict(X)
    wcss.append(pipeline['kmeans'].inertia_)

plt.plot(wcss)
```

```python
scaler = StandardScaler()
pipeline = Pipeline([
        ('preprocess',scaler),
        ('kmeans',KMeans(n_clusters=5,random_state=42,init='k-means++'))
    ])
pipeline.fit(X)
labels = pipeline['kmeans'].labels_
scaled_centroids = pipeline.named_steps['kmeans'].cluster_centers_
centroids = scaler.inverse_transform(scaled_centroids)

# plotting
plt.scatter(X['Annual Income (k$)'], X['Spending Score (1-100)'], c=labels, cmap='viridis', marker='o')
plt.scatter(centroids[:,0], centroids[:,1], c='red', marker='x', s=200, label='Centroids')
plt.xlabel('Annual Income (k$)')
plt.ylabel('Spending Score (1-100)')
plt.title('Mall Customers Clustering (k=5)')
plt.legend()
plt.show()
```
